```python
import nltk
from nltk.corpus import wordnet

nltk.download("wordnet")

word1 = "dog"
word2 = "cat"

synset1 = wordnet.synset(word1 + ".n.01")
synset2 = wordnet.synset(word2 + ".n.01")

similarity = synset1.wup_similarity(synset2)
print(f"Similarity between '{word1}' and '{word2}': {similarity:.2f}")  #.2f alow 2 digit after dot and prefix the number
```

```python
import nltk
from nltk.corpus import wordnet

# Download WordNet data (if not already downloaded)
nltk.download("wordnet")

# Define a word
word = "car"

# Retrieve synsets (sets of synonyms) for the word
synsets = wordnet.synsets(word)
print(synsets)

if synsets:
    # Find hyponyms (subordinate terms) for the first synset
    hyponyms = synsets[0].hyponyms()
    print('this: ',hyponyms)
    if hyponyms:
        print(f"Hyponyms for '{word}': {[hyponym.name().split('.')[0] for hyponym in hyponyms]}")

    # Find hypernyms (superordinate terms) for the first synset
    hypernyms = synsets[0].hypernyms()
    if hypernyms:
        print(f"Hypernyms for '{word}': {[hypernym.name().split('.')[0] for hypernym in hypernyms]}")
else:
    print(f"No synsets found for '{word}'")
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download("wordnet")

def get_synonyms(word):
    synonyms = []
    for synset in wordnet.synsets(word):
        synonyms.extend(synset.lemma_names())
    return set(synonyms)

word = "happy"
synonyms = get_synonyms(word)
print(f"Synonyms for '{word}' are: {', '.join(synonyms)}")
```

```python
import nltk
from nltk.corpus import wordnet

nltk.download("wordnet")

def get_antonyms(word):
    antonyms = []
    for synset in wordnet.synsets(word):
        for lemma in synset.lemmas():
            for antonym in lemma.antonyms():
                antonyms.append(antonym.name())
    return set(antonyms)

word = "hot"
antonyms = get_antonyms(word)
print(f"Antonyms for '{word}' are: {', '.join(antonyms)}")
```

```python
import nltk
from nltk.corpus import wordnet
from nltk.wsd import lesk

# Download WordNet data (if not already downloaded)
nltk.download("wordnet")

# Define two words
word1 = "dog"
word2 = "cat"

# Get the most common synset for each word using Lesk algorithm
synset1 = lesk(word1, word2)
synset2 = lesk(word2, word1)

if synset1 and synset2:
    # Calculate Wu-Palmer Similarity
    similarity = synset1.wup_similarity(synset2)
    if similarity is not None:
        print(f"Similarity between '{word1}' and '{word2}': {similarity:.2f}")
    else:
        print("No similarity found.")
else:
    print("Could not find synsets for both words.")
```

```python
import nltk
from nltk.corpus import wordnet

# Download WordNet data (if not already downloaded)
nltk.download("wordnet")

# Define a word
word = "good"

# Retrieve synsets (sets of synonyms) for the word
synsets = wordnet.synsets(word)

if synsets:
    # Print synonyms for the first synset
    synonyms = synsets[0].lemmas()
    print(f"Synonyms for '{word}': {[lemma.name() for lemma in synonyms]}")

    # Find antonyms for the first synset
    antonyms = synonyms[0].antonyms()
    if antonyms:
        print(f"Antonyms for '{word}': {[antonym.name() for antonym in antonyms]}")
    else:
        print(f"No antonyms found for '{word}'")
else:
    print(f"No synsets found for '{word}'")
```
